```python
for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print(col, "Outliers:", len(outliers))
```

# Data Cleaning and Preprocessing Documentation

## 1. Missing Values
Missing values were checked using `isnull().sum()`. Numerical missing values were replaced with the median because the dataset contains skewed numerical variables. Median imputation is less affected by extreme values than mean imputation.

## 2. Duplicate Records
Duplicate records were checked using `duplicated().sum()`.

## 3. Data Types
The numerical columns were converted into appropriate numeric data types. Invalid nonnumeric values were converted into `NaN` so that they could be handled during the missing-value treatment.

## 4. Categorical Consistency
The categorical columns were checked for inconsistent formatting. Extra spaces and differences in capitalization were standardized. The `planet_status` column was found to contain only the category `Confirmed`.

## 5. Invalid Values
Negative values were checked because measurements such as mass and radius should not be negative. Infinite values were also checked and replaced with missing values before performing statistical analysis.

## 6. Outliers
Outliers were detected using the Interquartile Range (IQR) method. A large number of outliers were found in several numerical columns. Since the dataset contains astronomical measurements, these extreme values may represent genuine observations. Therefore, the outliers were retained instead of being deleted.

## 7. Data Transformation
The numerical variables were examined for skewness. Log transformation can be applied to highly skewed variables such as mass, minimum mass, and radius to reduce the effect of extreme values and improve the distribution.

## 8. Statistical Analysis
Descriptive statistics were calculated, including mean, median, minimum, maximum, range, variance, and standard deviation. Correlation analysis was performed to understand relationships between numerical variables. Distribution analysis and statistical tests were also performed.

## 9. Final Validation
After cleaning, the final dataset was prepared for further data analysis, visualization, and machine learning.
